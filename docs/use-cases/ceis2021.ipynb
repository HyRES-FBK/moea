## Algorithm architecture

The functions stack in the
[original code](https://github.com/shaikatcse/EnergyPLANDomainKnowledgeEAStep1/blob/master/src/reet/fbk/eu/OptimizeEnergyPLANCEIS2021/Problem/EnergyPLANProblemCEIS2021V1Spooling.java)
by Mahbub worked as follows.

In the problem constructor, the following methods are called:

1. Extract the EnergyPLAN parameters, which retrives parameters for the target
scenario.
2. Read heat demand and solar thermal production distributions hour by hour.

In the problem evaluation method, the following function calls are executed:

1. **Creation of modification files**, that is the files containing the input
variables.
    1. Repair heat variables
    1. Repair transport variables
    1. Write variables into input files
1. **Simulate all scenarios** calls EnergyPLAN in spool mode.
1. **Extract information from solution files** and retrieve data to calculate
objectives and constraints.


```python
from pymoo.optimize import minimize

from moea.models import get_model
from moea.algorithms import get_algorithm


model_name = "ceis2021"
algorithm_name = "NSGA-II"
pop_size = 100
years = [2030, 2050]

results = []

for year in years:
    model = get_model(model_name, year=year)

    algorithm = get_algorithm(algorithm_name, pop_size=pop_size)

    results[year] = minimize(
        model,
        algorithm,
        ("n_gen", 100),
        seed=1,
        verbose=True,
    )
    break
```
